## General Preprocessing

### Preprocessing few continuous variables

```python
#Comando para conocer los diferentes valores de los datos contenidos en alguna columna del dataframe
loan_data["emp_length"].unique()
```

```python
#Comando para convertir remplazar un string: str.replace(valor_a_reemplazar, nuevo_valor)
loan_data["emp_length_int"] = loan_data["emp_length"].str.replace('+ years',"")
loan_data["emp_length_int"] = loan_data["emp_length_int"].str.replace('< 1 year',str(0))
loan_data["emp_length_int"] = loan_data["emp_length_int"].str.replace('nan',str(0))
loan_data["emp_length_int"] = loan_data["emp_length_int"].str.replace('years',"")
loan_data["emp_length_int"] = loan_data["emp_length_int"].str.replace('year', "")
```

```python
#Comando para saber el tipo de dato que quedo tras el reemplazo de las cadenas de texto
#La idea es que tras acortar el texto sea posible pasar de type string a type int

type(loan_data["emp_length_int"][0]) #Tipo de valor del primer elemento de la columna
```

```python
loan_data["emp_length_int"].dtype #Tipo de valores generales de la columna
```

```python
#Comando para forzar la conversion de strings a valores numericos
loan_data["emp_length_int"]=pd.to_numeric(loan_data["emp_length_int"])
```

```python
loan_data["emp_length_int"].dtype #Tipo de valores generales de la columna
```

```python
#Se define la manera para asignar valores Dummies (0 y 1) a las categorias que denoten impago, para ello utilizamos np.where(condition,value if true, value if false)   
# y el método .isin() para comprobar si un valor está en un alista de valores

loan_data["good_bad"] = np.where(loan_data["loan_status"].isin(["Charged Off", "Default", "Late (31-120 days)", "Does not meet the credit policy. Status:Charged Off" ]), 0, 1)
```

### Splitting Data

```python
#Usaremos la libreria Sklearn para poder hacer la división entre el conjunto de entrenamiento y el conjunto de testeo
from sklearn.model_selection import train_test_split
```

```python
#Sacamos del conjunto de entrenamiento la variable que queremos predecir (Good Bad): train_test_split(Dataset_entreno, variable a predecir)
train_test_split(loan_data.drop("good_bad", axis=1), loan_data["good_bad"])
#Se generan 4 valores: Columnas entrada filas de entrenamiento, Columnas entrada filas de prueba, filas de entranmiento variable a predecir, filas de prueba variable a predecir
```

```python
loan_data_inputs_train, loan_data_inputs_test, loan_data_targets_train, loan_data_targets_test = train_test_split(loan_data.drop("good_bad", axis=1), loan_data["good_bad"])
#Realizamos la asignación del desempaquetado de la funcion train_test_split()
```

```python
#Por defecto Sklearn asigna la proporcion 75/25 (Train/test), veamos otras maneras de cambiar esta proporcion: train_test_split(Dataset,variable_predecir, tamaño_conjunto_prueba)
#Dado que la forma en como se realiza la división es aleatoria, podemos usar una semilla para fijar la manera en como se realizó dicha elección a traves de random_state = 42
loan_data_inputs_train, loan_data_inputs_test, loan_data_targets_train, loan_data_targets_test = train_test_split(loan_data.drop("good_bad", axis=1), loan_data["good_bad"], test_size = 0.2, random_state =42)
#Proporcion 80/20
```
