Buscamos calular "Weight of Evidence (WoE)" de la variable discreta "Grade". Recordemos que $WoE = ln(\frac{\%good}{\%bad})$

Contamos cuántos préstamos hay en cada categoría de `grade`.

```python
df1.groupby(df1.columns.values[0], as_index =False)[df1.columns.values[1]].count()
```

Calculamos la proporción de préstamos buenos en cada categoría de `grade`. Como `good_bad` toma valores 0 y 1, su promedio equivale a esa proporción:

$$p_j = \frac{1}{N_j}\sum_{i \in j} good\_bad_i$$

donde $N_j$ es el número de préstamos del grade $j$.

```python
df1.groupby(df1.columns.values[0], as_index =False)[df1.columns.values[1]].mean()
```

```python
df1 = pd.concat([df1.groupby(df1.columns.values[0], as_index =False)[df1.columns.values[1]].count(),
               df1.groupby(df1.columns.values[0], as_index =False)[df1.columns.values[1]].mean()], axis =1)
```

```python
df1
```

```python
df1 = df1.iloc[:, [0,1,3]]
```

```python
df1
```

```python
df1.columns = [df1.columns.values[0], "n_obs", "prop_good"]
df1
```

Calculamos la proporción de observaciones por calificación:

$$\%Obs_j = \frac{N_j}{\sum_{j=1}^{K} N_j}$$

donde $N_j$ es `n_obs` del grade $j$ y $K$ es el número de categorías.

```python
df1["prop_n_obs"] = df1["n_obs"] /df1["n_obs"].sum()
```

```python
df1
```

Ahora calculamos el número de prestatarios buenos y el número de prestatarios malos por grupo de grado:

$$Good_j = p_j \cdot N_j \qquad\qquad Bad_j = (1 - p_j) \cdot N_j$$

donde $p_j$ es `prop_good` y $N_j$ es `n_obs` del grade $j$.

```python
df1["n_good"] = df1["prop_good"]*df1["n_obs"]
df1["n_bad"] = (1 - df1["prop_good"])*df1["n_obs"]
df1
```

Sacamos la proporción de buenos y de malos prestatarios de cada grade respecto al total de buenos y de malos:

$$\%Good_j = \frac{Good_j}{\sum_{j=1}^{K} Good_j} \qquad\qquad \%Bad_j = \frac{Bad_j}{\sum_{j=1}^{K} Bad_j}$$

```python
df1["prop_n_good"] = df1["n_good"] / df1["n_good"].sum()
df1["prop_n_bad"] = df1["n_bad"] / df1["n_bad"].sum()
df1
```

Calculamos el coeficiente de **Weight of Evidence (WoE)** de cada categoría:

$$WoE_j = \ln\left(\frac{\%Good_j}{\%Bad_j}\right)$$

Un WoE positivo indica que la categoría concentra más buenos que malos; uno negativo, lo contrario.

```python
df1["WoE"] = np.log(df1["prop_n_good"] / df1["prop_n_bad"])
df1
```

Veamos cuáles son las categorías con mayor tasa de impago a partir del WoE. Ordenamos de menor a mayor WoE (las categorías más riesgosas quedan primero) y reiniciamos el índice con `reset_index(drop=True)`.

```python
df1 = df1.sort_values(["WoE"])
df1 = df1.reset_index(drop = True)
df1
```

Calculemos las diferencias, en valor absoluto, entre las proporciones de buenos prestatarios de cada categoría y la anterior:

$$\Delta p_j = \left| p_j - p_{j-1} \right|$$

```python
df1["diff_prop_good"] = df1["prop_good"].diff().abs()
df1
```

Calculemos las diferencias, en valor absoluto, entre el WoE de cada categoría y el de la anterior:

$$\Delta WoE_j = \left| WoE_j - WoE_{j-1} \right|$$

```python
df1["diff_WoE"] = df1["WoE"].diff().abs()
df1
```

##### Valor Informativo (IV)
Recordemos que el Valor informativo nos indica cuanta infrmación aporta la variable independiente con respecto a la explicación de la variable dependiente y su capacidad de predicción.
Podemos Calcular IV a partir de la siguiente formula:

$$IV = \sum_{j=1}^{K} (\%Good_j - \%Bad_j) * WoE_j$$


| Rango: 0 - 1 | Poderes predictivos |
| :--- | :--- |
| **IV < 0.02** | Sin poder predictivo |
| **0.02 < IV < 0.1** | Poder predictivo débil |
| **0.1 < IV < 0.3** | Poder predictivo medio |
| **0.3 < IV < 0.5** | Poder predictivo fuerte |
| **0.5 < IV** | Sospechosamente alto, demasiado bueno para ser verdad |

```python
df1["IV"] = (df1["prop_n_good"] - df1["prop_n_bad"] )* df1["WoE"]
df1["IV"] = df1["IV"].sum()
df1
```

### Preprocessing Discrete Variables

Queremos automatizar el proceso de calcular el WoE para variables categóricas. Como anteriormente evidenciamos que se trata de un proceso mecánico y, por lo tanto, automatizable, podemos hacerlo a partir de una **función**.

La función `woe_discrete()` condensa en un solo bloque todos los pasos que realizamos antes, uno por uno, para la variable `grade`:

`woe_discrete(df, discrete_variable_name, good_bad_variable_df)`

| Parámetro | Descripción |
| :--- | :--- |
| `df` | DataFrame con las variables independientes (entradas) |
| `discrete_variable_name` | Nombre de la variable discreta a analizar (por ejemplo, `'grade'`) |
| `good_bad_variable_df` | Variable objetivo (`good_bad`) |

Los pasos que realiza la función son:

1. Unir la variable discreta con la variable objetivo.
2. Contar los préstamos por categoría (`n_obs`) y calcular la proporción de buenos (`prop_good`).
3. Calcular la proporción de observaciones por categoría (`prop_n_obs`).
4. Calcular el número de prestatarios buenos y malos (`n_good`, `n_bad`).
5. Calcular la proporción de buenos y malos sobre sus respectivos totales (`prop_n_good`, `prop_n_bad`).
6. Calcular el WoE de cada categoría:

$$WoE_j = \ln\left(\frac{\%Good_j}{\%Bad_j}\right)$$

7. Ordenar las categorías de menor a mayor WoE y reiniciar el índice.
8. Calcular las diferencias entre categorías consecutivas (`diff_prop_good`, `diff_WoE`).
9. Calcular el Valor Informativo (IV) de la variable:

$$IV = \sum_{j=1}^{K} (\%Good_j - \%Bad_j) \cdot WoE_j$$

La función devuelve un DataFrame con toda esta información, listo para decidir qué categorías conviene agrupar.

```python
def woe_discrete(df, discrete_variable_name, good_bad_variable_df):
    # 1. Unir la variable discreta con la variable objetivo
    df = pd.concat([df[discrete_variable_name], good_bad_variable_df], axis = 1)
    
    # 2. Contar los préstamos por categoría (n_obs) y calcular la proporción de buenos (prop_good)
    df = pd.concat([df.groupby(df.columns.values[0], as_index = False)[df.columns.values[1]].count(),
                    df.groupby(df.columns.values[0], as_index = False)[df.columns.values[1]].mean()], axis = 1)
    df = df.iloc[:, [0, 1, 3]]
    df.columns = [df.columns.values[0], 'n_obs', 'prop_good']
    
    # 3. Calcular la proporción de observaciones por categoría
    df['prop_n_obs'] = df['n_obs'] / df['n_obs'].sum()
    
    # 4. Calcular el número de prestatarios buenos y malos
    df['n_good'] = df['prop_good'] * df['n_obs']
    df['n_bad'] = (1 - df['prop_good']) * df['n_obs']
    
    # 5. Calcular la proporción de buenos y malos sobre sus respectivos totales
    df['prop_n_good'] = df['n_good'] / df['n_good'].sum()
    df['prop_n_bad'] = df['n_bad'] / df['n_bad'].sum()
    
    # 6. Calcular el WoE de cada categoría
    df['WoE'] = np.log(df['prop_n_good'] / df['prop_n_bad'])
    
    # 7. Ordenar de menor a mayor WoE y reiniciar el índice
    df = df.sort_values(['WoE'])
    df = df.reset_index(drop = True)
    
    # 8. Calcular las diferencias entre categorías consecutivas
    df['diff_prop_good'] = df['prop_good'].diff().abs()
    df['diff_WoE'] = df['WoE'].diff().abs()
    
    # 9. Calcular el Valor Informativo (IV) de la variable
    df['IV'] = (df['prop_n_good'] - df['prop_n_bad']) * df['WoE']
    df['IV'] = df['IV'].sum()
    
    return df
```

Ahora realizaremos el proceso de visualización para poder evidenciar el comportamiento del WoE y así sacar mejores conclusiones.

Usamos dos librerías:

- `matplotlib.pyplot` (`plt`): librería base para crear gráficos.

```python
import matplotlib.pyplot as plt
import seaborn as sns
sns.set()
```

La función `plot_by_woe()` grafica el WoE de cada categoría de una variable discreta. Recibe el DataFrame que devuelve `woe_discrete()`:

`plot_by_woe(df_WoE, rotation_of_x_axis_labels=0)`

| Parámetro | Descripción |
| :--- | :--- |
| `df_WoE` | DataFrame con el WoE por categoría (resultado de `woe_discrete()`) |
| `rotation_of_x_axis_labels` | Ángulo de rotación de las etiquetas del eje X (por defecto `0`). Es útil subirlo cuando hay muchas categorías o los nombres son largos |

Los pasos que realiza la función son:

1. Definir el eje X: los nombres de las categorías, que están en la primera columna, convertidos a texto con `apply(str)` para que se traten como etiquetas y no como números.
2. Definir el eje Y: el valor del WoE de cada categoría.
3. Crear la figura con `plt.figure(figsize=(ancho, alto))`. Aquí usamos `(18, 6)`.
4. Graficar la línea con `plt.plot()`: marcadores circulares (`marker="o"`), línea punteada (`linestyle="--"`) y color negro (`color="k"`).
5. Poner las etiquetas: el nombre de la variable en el eje X, "Weight of Evidence" en el eje Y y un título que incluye el nombre de la variable.
6. Rotar las etiquetas del eje X según el parámetro `rotation_of_x_axis_labels`.

Como el WoE se calcula por categoría y las categorías están ordenadas de menor a mayor WoE, la línea nos permite ver cómo cambia el comportamiento de una categoría a otra y detectar cuáles se parecen lo suficiente como para agruparse.

```python
def plot_by_woe(df_WoE, rotation_of_x_axis_labels=0):
    x =np.array(df_WoE.iloc[:,0].apply(str))
    y = df_WoE["WoE"]
    plt.figure(figsize= (18,6)) #figsize = (width, length)
    plt.plot(x,y, marker ="o", linestyle = "--", color = "k")
    plt.xlabel(df_WoE.columns[0])
    plt.ylabel("Weight of Evidence")
    plt.title(str("Weight of Evidence by" + df_WoE.columns[0]))
    plt.xticks(rotation = rotation_of_x_axis_labels)
```

Combinamos las categorías `RENT`, `OTHER`, `NONE` y `ANY` de `home_ownership` en una sola variable dummy, `RENT_OTHER_NONE_ANY`. `OTHER`, `NONE` y `ANY` tienen muy pocas observaciones y su WoE es poco confiable por separado (en `ANY`, el WoE es `inf` porque solo hay un préstamo y es bueno). `RENT`, aunque sí tiene muchas observaciones, comparte con ellas un WoE negativo, es decir, un comportamiento de pago peor que el de `OWN` y `MORTGAGE`.

Como las dummies son excluyentes, sumarlas da 1 si el prestatario pertenece a cualquiera de estas categorías y 0 en caso contrario.

```python
df_inputs_prepr["home_ownership:RENT_OTHER_NONE_ANY"] = sum([df_inputs_prepr["home_ownership:RENT"], df_inputs_prepr["home_ownership:OTHER"], 
                                                            df_inputs_prepr["home_ownership:NONE"], df_inputs_prepr["home_ownership:ANY"]])
```
